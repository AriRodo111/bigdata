### Actividad 3 — Estadística descriptiva

Cálculo de las medidas de tendencia central y de dispersión sobre la variable `Annual change in population`, para el subconjunto de **Granada (Grenada) entre 1983 y 2021**.


```python
df_ext_EY= df[ (df["Year"] >= 1983) & (df["Year"] <= 2021) & (df["Entity"] == "Grenada") ]
df_ext_EY.head(5)
```

```python
df_ext_EY.mean(numeric_only=True)
```

```python
df_ext_EY.median(numeric_only=True)
```

```python
df_ext_EY.var(numeric_only=True)
```

```python
df_ext_EY.std(numeric_only=True)
```

```python
df_ext_EY.min(numeric_only=True)
```

```python
df_ext_EY.max(numeric_only=True)
```

```python
df.min(numeric_only=True)
```

```python
df.max(numeric_only=True)
```

### Interpretación

Para **Granada entre 1983 y 2021**, el cambio anual de población presenta los siguientes valores:

| Medida | Valor |
|--------|-------|
| Media | 571.72 personas por año |
| Mediana | 564.00 personas por año |
| Varianza | 32 624.63 |
| Desviación estándar | 180.62 personas |
| Mínimo | 255 personas |
| Máximo | 840 personas |

La **media (571.72)** indica que, en promedio, Granada ganó alrededor de 572 habitantes por año en ese periodo. Como la **mediana (564)** es prácticamente igual a la media, la distribución es **aproximadamente simétrica**: no hay años atípicos que distorsionen el promedio.

La **desviación estándar (180.62)** mide cuánto se separan los años individuales de ese promedio; comparada con una media de 571.72 es una dispersión moderada, lo que refleja un crecimiento sostenido pero irregular. La **varianza (32 624.63)** es esa misma dispersión al cuadrado, por lo que no se expresa en personas y se usa sobre todo como paso intermedio.

El **mínimo (255)** y el **máximo (840)** muestran que en todos los años del periodo el cambio fue **positivo**: Granada nunca perdió población entre 1983 y 2021, solo creció más rápido en unos años que en otros.

Al comparar con el dataset completo (`df.min()` y `df.max()`), el rango global va de **-3 315 935** hasta **93 371 370** personas: los valores negativos corresponden a entidades que perdieron población y los máximos a agregados mundiales. Esto confirma la enorme diferencia de escala entre un país pequeño como Granada y los agregados.
